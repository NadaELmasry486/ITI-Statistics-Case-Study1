```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_clean = pd.read_excel("cleaned_data.xlsx")
```

```python
# Count students in each Branch × Track stratum
strata = (
    df_clean
    .groupby(["branch", "track"])
    .size()
    .reset_index(name="count")
)

# Calculate each stratum's proportion
strata["proportion"] = strata["count"] / len(df_clean)

# Proportional allocation for a sample of 200 students
sample_size = 200

strata["sample_size"] = (
    strata["proportion"] * sample_size
).round().astype(int)

strata
```
