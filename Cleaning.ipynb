### converting DataTypes
#### Identifier

```python
df_clean["student_id"] = df_clean["student_id"].astype("string")
```

#### Qualitative — Nominal

```python
nominal_cols = {
    "branch":         ["Cairo", "Alexandria", "Mansoura", "Assiut", "Menoufia"],
    "track":          ["Data Science", "Web Development", "Cyber Security",
                       "Embedded Systems", "UI/UX"],
    "gender":         ["Male", "Female"],
    "has_laptop":     ["Yes", "No"],
    "transport":      ["Bus", "Metro", "Car", "Walk", "Microbus"],
    "favorite_sport": ["Football", "Swimming", "Basketball",
                       "Running", "Volleyball", "None"],
}
for col, order in nominal_cols.items():
    df_clean[col] = pd.Categorical(df_clean[col], categories=order, ordered=False)
```

#### Qualitative — Ordinal

```python
df_clean["satisfaction_1to5"] = pd.Categorical(
    df_clean["satisfaction_1to5"],
    categories=[1, 2, 3, 4, 5],
    ordered=True,
)

df_clean["satisfaction_1to5"].value_counts(dropna=False).sort_index()
```

#### Quantitative — Discrete

```python
for col in ["age", "projects_submitted", "pets_at_home"]:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
    df_clean[col] = df_clean[col].astype("Int64")
```

#### Quantitative — Continuous (float)

```python
for col in ["study_hours_week", "commute_minutes", "sleep_hours", "final_score"]:
    df_clean[col] = df_clean[col].astype("float64")
```
